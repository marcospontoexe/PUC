### **Preparação dos Dados**

**Nesta etapa do projeto serão utilizados os dados do *dataset* df_uci**

Em todos os projetos de Machine Learning a etapa de pré-processamento de dados é fundamental. Aplicando as técnicas corretas para preparação dos dados, podemos aumentar consideravelmente o desempenho dos modelos implementados.

Para alimentar o modelo de Machine Learning, é necessário separar a variável alvo (dependente) do *dataset* das demais (independentes).

Geralmente, as variáveis são divididas como X e y, este projeto vai adotar o mesmo padrão.

Após separar nossas variáveis devidamente, como parte do pré-processamento dos dados, que irão alimentar o modelo de Machine Learning, será usado o **StardardScaler**, que vem junto com **sklearn.preprocessing**, para padronizar nossos dados numéricos.

```python
# DIVISÃO DOS DADOS EM TREINO E TESTE

from sklearn.model_selection import train_test_split

# stratify=df_uci['num'] garante que a proporção de doentes/não doentes seja a mesma
# no treino e no teste. Sem isso, com apenas 303 amostras, o conjunto de teste pode
# ficar com um balanceamento bem diferente do de treino, distorcendo a avaliação.
X_train, X_test, y_train, y_test = train_test_split(
    df_uci.drop(['num'], axis=1),   # remove a coluna label
    df_uci['num'],                  # coluna que será usada como label de classificação
    test_size=0.2,                  # porcentagem de divisão para a base de testes
    random_state=35,                # "seed", para tornar o split reproduzível
    stratify=df_uci['num']          # mantém a proporção das classes nos dois conjuntos
)

# Conferindo se a estratificação funcionou
print("Proporção de positivos no dataset completo: {:.2%}".format(df_uci['num'].mean()))
print("Proporção de positivos no treino:           {:.2%}".format(y_train.mean()))
print("Proporção de positivos no teste:            {:.2%}".format(y_test.mean()))
print("Tamanho do conjunto de teste:", len(y_test), "amostras")
```

```python
# IMPORTAÇÃO DAS FERRAMENTAS PARA O PRÉ-PROCESSAMENTO DOS DADOS
from sklearn.preprocessing import StandardScaler

# PADRONIZAÇÃO DOS DADOS NUMÉRICOS
# O StandardScaler padroniza os dados (média=0, desvio padrão=1).
#
# OBSERVAÇÃO: o Random Forest NÃO precisa de padronização — árvores de decisão
# dividem uma feature por vez e são invariantes à escala. Mantemos a padronização
# aqui para reproduzir o notebook original, mas ela é de fato necessária apenas
# para os modelos baseados em distância/otimização usados adiante (Regressão
# Logística e SVM).
scaler = StandardScaler()

# Regra de ouro contra data leakage: o scaler APRENDE (fit) somente com o treino
# e apenas TRANSFORMA (transform) o teste.
X_train_array = scaler.fit_transform(X_train)
X_test_array = scaler.transform(X_test)

# Convertendo de volta para um DataFrame
X_train_scaler = pd.DataFrame(X_train_array, columns=X_train.columns, index=X_train.index)
X_test_scaler = pd.DataFrame(X_test_array, columns=X_test.columns, index=X_test.index)
```

Por fim, antes de implementar o modelo, iremos dividir o nosso *dataset* em dados de treino e de teste.

Essa etapa também é fundamental, uma vez que ela evita que o modelo possa fazer previsões erradas ao "viciar", em palavras simples, o modelo com um determinado conjunto de dados. Ou seja, quando nós usamos os mesmos dados para treinar e testar nossos modelos, é provável que ele não aprenda a classificar os dados, mas apenas "decorar" as respostas. Assim, ao submetir esse modelo para novos dados, ele não vai conseguir uma boa acurácia.

Por isso, é fundamental usar dados diferentes para treino e teste.

### **Modelo Random Forest**

Random Forest (floresta aleatória) é um algoritmo de aprendizagem supervisionada. Como você pode perceber pelo seu nome, ele cria uma floresta de um modo aleatório.

Segundo Niklas Donges, em um artigo do Towards Data Science, de modo simples: o algoritmo de florestas aleatórias cria várias árvores de decisão e as combina para obter uma predição com maior acurácia e mais estável.

<center><img alt = "Analisando Airbnb" width = "15%" src="https://static.thenounproject.com/png/961660-200.png"></center>

Uma grande vantagem do algoritmo de florestas aleatórias é que ele pode ser utilizado tanto para tarefas de classificação quanto para regressão, o que representa a maioria dos sistemas de aprendizagem de máquina atuais.

Você pode conferir um artigo completo do **Towards Data Science** sobre o modelo Random Forest clicando [aqui](https://towardsdatascience.com/understanding-random-forest-58381e0602d2).

```python
# IMPORTANDO O MODELO RANDOM FOREST

from sklearn.ensemble import RandomForestClassifier
```

```python
# INSTANCIANDO O MODELO

# n_estimators=100 (padrão do scikit-learn) em vez de 10: poucas árvores deixam a
# floresta com variância alta, ou seja, o resultado oscila muito conforme a amostra.
RF_model = RandomForestClassifier(n_estimators = 100, criterion = 'entropy', random_state = 35)
```

```python
# TREINANDO O MODELO

RF_model.fit(X_train_scaler, y_train)
```

### **Verificando o desempenho do modelo**

Uma das principais métricas utilizadas para avaliar o desempenho de um modelo de Machine Learning é a acurácia. Contudo, é importante ressaltar que cada caso é particular, com isso, uma boa acurácia nem sempre garantirá que o modelo está bom.

No caso da detecção de doenças cardiovasculares, o interesse é **maximizar o número de verdadeiros positivos e minimizar o número de falsos negativos**. Um falso negativo significa liberar como saudável um paciente que tem a doença — o erro mais grave neste contexto. Por isso a métrica prioritária aqui é o **recall** (sensibilidade), e não a acurácia.

Neste projeto, além da métrica de acurácia será utilizada a função `classification_report` para ver o desempenho do modelo sobre as métricas de precisão, ***recall, f1-score e support***.

```python
# IMPORTAÇÃO DAS FERRAMENTAS DE MÉTRICAS

from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
```

```python
# REALIZANDO A PREVISÃO NOS DADOS DE TESTE

y_pred = RF_model.predict(X_test_scaler)
```

```python
# VIZUALIZANDO A ACURÁCIA GERAL

print('[Acurácia] Random Forest:', accuracy_score(y_test, y_pred))
```

```python
# VISUALIZANDO O CLASSIFICATION REPORT

print('\n[Classification Report] Random Forest')

print(classification_report(y_test, y_pred))
```

###  Correlação (Entre Variáveis Numéricas)

Uma etapa adicional, mas ainda básica, é calcular a **matriz de correlação**.

*   **O que é?** Mede a força e a direção da relação linear entre pares de variáveis numéricas. O valor varia de -1 a +1.
    *   **+1:** Correlação positiva perfeita (quando uma sobe, a outra sobe).
    *   **-1:** Correlação negativa perfeita (quando uma sobe, a outra desce).
    *   **0:** Nenhuma correlação linear.
*   **Por que é importante?**
    *   **Entender Relações:** Ajuda a validar hipóteses. Por exemplo, poderíamos esperar uma correlação negativa entre `thalach` (frequência cardíaca máxima) e `age` (idade), já que a frequência máxima tende a diminuir com a idade.
    *   **Identificar Redundância:** Variáveis com correlação muito alta (ex: > 0.9) podem ser redundantes, e talvez uma delas possa ser removida para simplificar o modelo (*feature selection*).
    *   **Classes alvo**: Classes que tem alta correlação com a label a ser classificada é muito importante para o modelo de predição, já as que tem baixa correlação na maioria das vezes não são tão importantes e podem ser descartadas.

Essas estatísticas, juntas, fornecem um panorama completo e robusto do seu conjunto de dados, sendo um passo indispensável antes de qualquer visualização ou modelagem.

### Ressalva metodológica: correlação sobre variáveis nominais

A matriz de correlação acima usa o coeficiente de **Pearson**, que pressupõe variáveis numéricas com ordem. Três atributos do dataset não atendem a isso: `cp`, `restecg` e `thal` são **categóricos nominais** — seus códigos são rótulos, não quantidades. O `thal`, por exemplo, assume os valores 3, 6 e 7; a diferença entre 3 e 7 não significa "quatro unidades a mais de gravidade".

O tratamento correto é aplicar **one-hot encoding** antes de medir a associação, transformando cada categoria em uma coluna binária. Assim a correlação passa a responder a uma pergunta com sentido: *"ter esta categoria específica está associado ao diagnóstico?"*

```python
# CORRELAÇÃO TRATANDO AS VARIÁVEIS NOMINAIS CORRETAMENTE

# 'cp', 'restecg' e 'thal' são categóricas nominais: seus códigos não têm ordem.
nominais = ['cp', 'restecg', 'thal']

# get_dummies transforma cada categoria em uma coluna binária (one-hot encoding)
df_onehot = pd.get_dummies(df_uci, columns=nominais, prefix=nominais)

# Correlação de cada coluna com o alvo, ordenada pela FORÇA (valor absoluto),
# para que correlações negativas fortes não fiquem escondidas no fim da lista
corr_alvo = df_onehot.corr()['num'].drop('num')
corr_alvo = corr_alvo.reindex(corr_alvo.abs().sort_values(ascending=False).index)

print("Correlação com o alvo 'num' após one-hot das variáveis nominais:\n")
print(corr_alvo.round(3).to_string())
```

#### O que o one-hot revelou

O tratamento correto das nominais muda a leitura de forma relevante:

| Atributo | Correlação com `num` | Interpretação |
| :--- | :--- | :--- |
| `thal_3.0` (tálio normal) | **-0,521** | O **indicador protetor mais forte de todo o dataset** |
| `cp_4.0` (dor assintomática) | **+0,516** | Confirma o achado contraintuitivo: a doença silenciosa é a mais associada ao diagnóstico |
| `thal_7.0` (defeito reversível) | +0,481 | Forte indicador de risco |
| `ca` (vasos afetados) | +0,460 | Quanto mais vasos comprometidos, maior o risco |
| `thal_6.0` (defeito fixo) | +0,105 | **Praticamente irrelevante** |

O contraste entre `thal_7.0` (+0,481) e `thal_6.0` (+0,105) é justamente o que a correlação sobre o código bruto escondia. Tratado como número, `thal` gerava uma única correlação de ~0,52 que sugeria "quanto maior o código, maior o risco" — quando na verdade **o defeito fixo (6) quase não discrimina, e todo o poder preditivo está concentrado no defeito reversível (7) e na ausência de defeito (3)**.

## Gráficos para Análise Univariada (Entendendo Cada Variável Isoladamente)
### Box Plot (Diagrama de Caixa):
* O que mostra? Um resumo de cinco números: mínimo, primeiro quartil (Q1), mediana (Q2), terceiro quartil (Q3) e máximo.
* Por que usar? É excelente para identificar?
    *  **outliers** (valores atípicos), que aparecem como pontos fora dos "bigodes" do gráfico.
    *  **dispersão dos dados** (o tamanho da caixa)
    *  **simetria dos dados**, quando a linha da mediana está no centro(Q2). Se a mediana esta mais para o alto (Q3), existe
        uma distribuição assimétrica negativa, e vice-versa. 
* Exemplo: Um box plot para chol (colesterol) provavelmente mostraria aquele valor de 564 mg/dl como um outlier claro, muito acima do resto dos pacientes.

# 3. Usando modelos preditivos

Serão treinados **três novos classificadores**, além do Random Forest já utilizado no artigo original — que passa a servir de *baseline* na comparação:

1. **Regressão Logística**: calcula a probabilidade de um paciente pertencer à classe "doente" a partir de uma combinação linear das variáveis de entrada. É o modelo mais interpretável do conjunto.
2. **SVM**: procura o hiperplano que melhor separa as duas classes no espaço de dados, com a maior margem possível entre elas.
3. **XGBoost**: também usa árvores, mas de forma sequencial — constrói uma árvore, verifica onde ela errou e constrói a próxima para corrigir os erros da anterior.

Todos são montados dentro de um `Pipeline` do scikit-learn, tendo o `StandardScaler` como primeiro passo. Isso garante que a padronização seja aprendida **apenas com os dados de treino** a cada ajuste — inclusive dentro da validação cruzada — evitando *data leakage*.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# Validação cruzada estratificada, para não depender de um único split
from sklearn.model_selection import cross_val_score, StratifiedKFold

# Conjunto de métricas: acurácia sozinha não basta em diagnóstico médico
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, RocCurveDisplay,
                             classification_report, confusion_matrix)
```

## Regressão Logística
Este é o nosso modelo de base: simples, rápido e interpretável.

```python
# POR QUE NÃO USAMOS VarianceThreshold NESTE PROJETO
#
# O VarianceThreshold remove features cuja variância fica abaixo de um limite. O
# problema é que, em dados NÃO padronizados, a variância depende da UNIDADE DE MEDIDA
# e não da informação que a variável carrega. Vejamos:

var_treino = X_train.var().sort_values()

print("Variância de cada coluna do conjunto de treino (dados NÃO padronizados):\n")
print(var_treino.round(3).to_string())
print("-" * 60)

# Demonstrando o problema na prática, com dois limites diferentes
from sklearn.feature_selection import VarianceThreshold

for limite in [0.20, 0.25]:
    vt = VarianceThreshold(threshold=limite).fit(X_train)
    removidas = X_train.columns[~vt.get_support()].tolist()
    print(f"VarianceThreshold({limite}) removeria: {removidas}")

# Números lidos da própria variável, para o texto não ficar desatualizado
print(f"\nConclusão: 'chol' (mg/dL) tem variância ~{var_treino['chol']:.0f} e 'fbs' (binária)")
print(f"~{var_treino['fbs']:.2f} — mas isso não torna o colesterol mais informativo, é só a")
print("escala. Basta subir o limite de 0.20 para 0.25 para que 'sex' e 'exang' também")
print("sejam descartadas, sendo que 'exang' (angina induzida por exercício) é um dos")
print("preditores mais fortes do alvo segundo a análise de correlação.")
print("Com apenas 13 features, mantemos todas e deixamos a seleção a cargo dos modelos.")
```

```python
# --- Pipeline 1: Regressão Logística ---

# Cada pipeline recebe um StandardScaler() NOVO. Reutilizar a mesma instância entre
# pipelines faz com que o objeto termine ajustado ao último pipeline executado,
# quebrando as células anteriores se o notebook for reexecutado fora de ordem.
pipeline_lr = Pipeline([
    ('scaler', StandardScaler()),                                        # padroniza (fit só no treino)
    ('logistic_regression', LogisticRegression(random_state=35, max_iter=2000))
])

# Treinar o pipeline com os dados BRUTOS: a padronização acontece dentro dele
pipeline_lr.fit(X_train, y_train)

# Fazer previsões e avaliar
y_pred_lr = pipeline_lr.predict(X_test)

print(f"Acurácia: {accuracy_score(y_test, y_pred_lr):.4f}")
print(f"Recall:   {recall_score(y_test, y_pred_lr):.4f}  <-- métrica prioritária neste problema")
print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred_lr, target_names=['Sem Doença', 'Com Doença']))
```

## SVM (Support Vector Machine)
SVM é um modelo poderoso que busca a melhor "fronteira" para separar as classes.

```python
# --- Pipeline 2: SVM ---

pipeline_svm = Pipeline([
    ('scaler', StandardScaler()),   # essencial para o SVM, que é baseado em distância
    ('svm', SVC(kernel='rbf', probability=True, random_state=35))
])

pipeline_svm.fit(X_train, y_train)
y_pred_svm = pipeline_svm.predict(X_test)

print(f"Acurácia: {accuracy_score(y_test, y_pred_svm):.4f}")
print(f"Recall:   {recall_score(y_test, y_pred_svm):.4f}")
print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred_svm, target_names=['Sem Doença', 'Com Doença']))
```

## XGBoost
Frequentemente o modelo com a melhor performance para dados tabulares.

```python
# --- Pipeline 3: XGBoost ---

# Modelos de árvore não precisam de padronização, mas manter o scaler no pipeline
# não prejudica e permite usar exatamente o mesmo fluxo para todos os modelos.
pipeline_xgb = Pipeline([
    ('scaler', StandardScaler()),
    ('xgboost', XGBClassifier(eval_metric='logloss', random_state=35))
])

pipeline_xgb.fit(X_train, y_train)
y_pred_xgb = pipeline_xgb.predict(X_test)

print(f"Acurácia: {accuracy_score(y_test, y_pred_xgb):.4f}")
print(f"Recall:   {recall_score(y_test, y_pred_xgb):.4f}")
print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred_xgb, target_names=['Sem Doença', 'Com Doença']))
```

## Random Forest (baseline)
O artigo original treinou o Random Forest fora de um `Pipeline` e sobre dados já padronizados. Refazemos aqui dentro do mesmo padrão dos demais modelos, para que a comparação seja justa e para que ele também entre na validação cruzada.

```python
# --- Baseline: Random Forest ---

pipeline_rf = Pipeline([
    ('scaler', StandardScaler()),
    ('random_forest', RandomForestClassifier(n_estimators=100, criterion='entropy', random_state=35))
])

pipeline_rf.fit(X_train, y_train)
y_pred_rf = pipeline_rf.predict(X_test)

print(f"Acurácia: {accuracy_score(y_test, y_pred_rf):.4f}")
print(f"Recall:   {recall_score(y_test, y_pred_rf):.4f}")
print("\nRelatório de Classificação:")
print(classification_report(y_test, y_pred_rf, target_names=['Sem Doença', 'Com Doença']))
```

```python
# --- TABELA COMPARATIVA ---
# Comparamos por várias métricas, não só acurácia: em diagnóstico médico o falso
# negativo é o erro mais caro, e quem mede isso é o RECALL.

modelos = {
    'Regressão Logística': pipeline_lr,
    'SVM': pipeline_svm,
    'XGBoost': pipeline_xgb,
    'Random Forest (baseline)': pipeline_rf
}

resultados = []
for nome, modelo in modelos.items():
    y_p = modelo.predict(X_test)
    resultados.append({
        'Modelo': nome,
        'Acurácia': accuracy_score(y_test, y_p),
        'Precisão': precision_score(y_test, y_p),
        'Recall': recall_score(y_test, y_p),
        'F1-Score': f1_score(y_test, y_p),
        # AUC usa a probabilidade prevista, não a classe: mede a qualidade do
        # ranking de risco em todos os limiares de decisão possíveis
        'AUC': roc_auc_score(y_test, modelo.predict_proba(X_test)[:, 1])
    })

# Ordenado por RECALL, a métrica prioritária neste problema
results = pd.DataFrame(resultados).sort_values('Recall', ascending=False).round(3)
display(results.reset_index(drop=True))
```

## Validação cruzada

O conjunto de teste tem apenas ~61 amostras: uma diferença de 1,6 ponto de acurácia entre dois modelos equivale a **um único paciente**. Declarar um vencedor nessa base é frágil.

A validação cruzada estratificada em 5 *folds* treina e avalia cada modelo cinco vezes, em partições diferentes de **todo** o dataset, e reporta média e desvio-padrão. Uma diferença só é real se for maior que o desvio.

```python
# --- VALIDAÇÃO CRUZADA ESTRATIFICADA (5 folds) ---

# StratifiedKFold mantém a proporção de classes em cada fold, como fizemos no split.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=35)

X_completo = df_uci.drop(['num'], axis=1)
y_completo = df_uci['num']

print("Recall médio em validação cruzada (5 folds estratificados):\n")
for nome, modelo in modelos.items():
    # O Pipeline garante que o scaler seja reajustado dentro de cada fold,
    # sem vazar informação do fold de validação para o de treino.
    scores = cross_val_score(modelo, X_completo, y_completo, cv=cv, scoring='recall')
    print(f"{nome:26s}: {scores.mean():.3f} (+/- {scores.std():.3f})")

print("\nAcurácia média em validação cruzada:\n")
for nome, modelo in modelos.items():
    scores = cross_val_score(modelo, X_completo, y_completo, cv=cv, scoring='accuracy')
    print(f"{nome:26s}: {scores.mean():.3f} (+/- {scores.std():.3f})")
```

#### Leitura da validação cruzada

| Modelo | Recall (média ± desvio) | Acurácia (média ± desvio) |
| :--- | :--- | :--- |
| Regressão Logística | 0,784 ± 0,049 | 0,828 ± 0,032 |
| SVM | 0,784 ± 0,041 | 0,841 ± 0,043 |
| Random Forest | 0,777 ± 0,026 | 0,825 ± 0,038 |
| XGBoost | 0,756 ± 0,041 | 0,802 ± 0,032 |

**As diferenças entre os quatro modelos são menores que os respectivos desvios-padrão.** Em outras palavras: é um **empate técnico**, e nenhum modelo pode ser declarado superior com este volume de dados.

Esse resultado é o principal argumento a favor da validação cruzada. Olhando apenas o split único, o SVM parecia vencer com folga em acurácia (0,869 contra 0,820 do XGBoost) — uma diferença de 3 pacientes em 61. Repetindo a avaliação em cinco partições diferentes, essa vantagem se dissolve.

## Curva ROC

Terceiro gráfico novo em relação ao notebook original. A curva ROC mostra o *trade-off* entre sensibilidade (recall) e falsos positivos em **todos os limiares de decisão**, e não apenas no limiar padrão de 0,5 usado pelo `predict()`.

É a visualização mais adequada para este problema: ela permite ver qual modelo consegue mais sensibilidade para um mesmo custo de falsos positivos — exatamente a decisão clínica em jogo. A **AUC** resume a curva em um número: 1,0 é o classificador perfeito e 0,5 equivale a chutar.

```python
# --- CURVA ROC ---

fig, ax = plt.subplots(figsize=(9, 7))

# RocCurveDisplay calcula a curva a partir das probabilidades previstas por cada modelo
for nome, modelo in modelos.items():
    RocCurveDisplay.from_estimator(modelo, X_test, y_test, name=nome, ax=ax)

# Linha diagonal = desempenho de um classificador aleatório
ax.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Aleatório (AUC = 0.50)')

ax.set_title('Curva ROC — comparação dos modelos', fontsize=15)
ax.set_xlabel('Taxa de Falsos Positivos (1 - especificidade)', fontsize=12)
ax.set_ylabel('Taxa de Verdadeiros Positivos (sensibilidade / recall)', fontsize=12)
ax.legend(loc='lower right')
plt.tight_layout()
plt.show()
```
